## anyplotlib

Fast, interactive browser-side plotting using [anywidget](https://anywidget.dev/) + canvas.
All rendering happens in the browser — no kernel round-trip per frame.

```python
import numpy as np
import anyplotlib as apl
```

### Basic line plot

```python
x = np.linspace(0, 4 * np.pi, 512)
signal = np.sin(x)

fig, ax = apl.subplots(1, 1, figsize=(620, 320))
v = ax.plot(signal, axes=[x], units="rad")
fig
```

### Image display with colormap

```python
data = np.random.rand(128, 128)
fig, ax = apl.subplots(1, 1)
ax.imshow(data, cmap="viridis")
fig
```

### Bar chart

```python
categories = ["A", "B", "C", "D", "E"]
values = np.random.randint(10, 100, len(categories))
fig, ax = apl.subplots(1, 1)
ax.bar(categories, values)
fig
```
